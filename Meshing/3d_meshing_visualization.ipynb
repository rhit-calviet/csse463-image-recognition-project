# Poisson Mesh

```python
# This code is intended to provide examples of the different 3d meshing outputs generated during testing. Generated with help from GPT 5.2.

import open3d as o3d
import numpy as np
import plotly.graph_objects as go

# ---- Load .ply file ----
file_path = "/work/csse463/202620/04/csse463-image-recognition-project/Meshing/new_reconstructed_poisson_1.ply"  # <-- change this
mesh = o3d.io.read_triangle_mesh(file_path)

# Compute normals if missing (optional but recommended)
if not mesh.has_vertex_normals():
    mesh.compute_vertex_normals()

# ---- Convert to numpy ----
vertices = np.asarray(mesh.vertices)
triangles = np.asarray(mesh.triangles)

# ---- Plot using your preferred style ----
fig = go.Figure(data=[
    go.Mesh3d(
        x=vertices[:,0],
        y=vertices[:,1],
        z=vertices[:,2],
        i=triangles[:,0],
        j=triangles[:,1],
        k=triangles[:,2],
        color='orange',
        opacity=1.0,
        lighting=dict(
            ambient=0.3,
            diffuse=0.8,
            specular=0.3,
            roughness=0.5
        )
    )
])

fig.update_layout(
    scene=dict(
        aspectmode='data',   # equal axis scaling
        xaxis=dict(visible=False),
        yaxis=dict(visible=False),
        zaxis=dict(visible=False),
        bgcolor='white'
    ),
    paper_bgcolor='white'
)

fig.show()
```
