# Baseline Model Training Comparison (20 Class-Classification)

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
alexnet_df = pd.read_csv('/work/csse463/202620/04/data/alexnet_20classes_training_data.csv')
resnet_df = pd.read_csv('/work/csse463/202620/04/data/resnet18_20classes_training_data.csv')
```

```python
# Plot loss

plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.plot(alexnet_df['epoch'], alexnet_df['train_loss'], marker='s', label='AlexNet')
plt.plot(resnet_df['epoch'], resnet_df['train_loss'], marker='^', label='ResNet18')
plt.xlabel('Epoch')
plt.ylabel('Training Loss')
plt.title('Training Loss per Epoch')
plt.grid(True)
plt.legend()
```

```python
# Plot Accuracy

plt.figure(figsize=(12,6))
plt.plot(alexnet_df['epoch'], alexnet_df['val_accuracy'], marker='s', label='AlexNet')
plt.plot(resnet_df['epoch'], resnet_df['val_accuracy'], marker='^', label='ResNet18')
plt.xlabel('Epoch')
plt.ylabel('Validation Accuracy (%)')
plt.title('Validation Accuracy per Epoch')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.savefig('model_comparison_loss_accuracy.png', dpi=300)
plt.show()
```
